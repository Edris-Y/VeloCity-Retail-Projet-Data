```python
import pandas as pd

# 1. Chargement de la donnée
# (Remplace par le bon chemin si ton fichier n'est pas dans le même dossier)
df = pd.read_csv('master_table_velocity.csv')

# 2. Transformation des dates (très important pour les séries temporelles plus tard)
df['order_date'] = pd.to_datetime(df['order_date'])

# 3. FEATURE ENGINEERING : Calcul du vrai chiffre d'affaires
# Formule : (Prix * Quantité) - La remise
# Si discount = 0.20 (20%), alors on multiplie par (1 - 0.20) = 0.80
df['final_revenue'] = (df['list_price'] * df['quantity']) * (1 - df['discount'])

# 4. Création du mois et de l'année pour faciliter l'analyse
df['order_year'] = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month

# 5. On affiche les 5 premières lignes pour vérifier notre travail !
display(df[['order_date', 'product_name', 'list_price', 'discount', 'final_revenue']].head())

df.to_csv('velocity_powerbi_ready.csv', index=False)

print("\n✅ Fichier 'velocity_powerbi_ready.csv' généré avec succès ! Prêt pour Power BI !")
```
